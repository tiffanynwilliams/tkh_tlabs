```python
print("*"* 25)
print("Analytical Questions - Q1")
print("*"* 25)

color_by_season = shopping_data.pivot_table(values = 'Purchase Amount (USD)', index='Color', columns='Season', aggfunc= 'count')
pop_color = color_by_season.idxmax()

print("The most popular colour by season are as follows-- Fall: Brown, Spring: Baby Blue, Summer: Lavender, and Winter: Black.")
```

```python
print("*"* 25)
print("Analytical Questions - Q2")
print("*"* 25)

item_by_season = shopping_data.pivot_table(values='Purchase Amount (USD)', index='Item Purchased', columns= 'Season', aggfunc='count')

most_pop_item = item_by_season.idxmax()
print(most_pop_item)


print("The following items are the most popular for each season-- Fall: Socks, Spring: Running Shoes, Summer = Shorts, Winters = Leggings.\n"
"Initially, there was concern for the NaNs appearing in the seasons, but upon further" \
" investigation, we can safely conclued that certain items\nwere not purchased during" \
" certain months (e.g., warmer clothing items like boots and scarves were NaN in the Spring " \
"and Summer season;\ncooler items like shorts and dresses were NaN in Fall and Winter season.)\n")
```

```python
print("*"* 25)
print("Analytical Questions - Q3")
print("*"* 25)

promo_by_usd = (round(shopping_data.groupby('Promo Code Used')['Purchase Amount (USD)'].mean(),2))
usd_max = promo_by_usd.max()
usd_min = promo_by_usd.min()

diff_usd = usd_max - usd_min

print(f'Customers spent ${diff_usd} more when they used promo codes.'
'\nMaybe there is a sort of buying bias where the customer thinks they are getting a better deal while they are spending more money in actuality.')
```

```python
print("*"* 25)
print("Analytical Questions - Q4")
print("*"* 25)

review_by_usd = round(shopping_data.groupby('Review Rating')['Purchase Amount (USD)'].mean(),2)
review_usd_max = review_by_usd.max()
review_usd_min = review_by_usd.min()

diff_review_usd = round((review_usd_max - review_usd_min),2)

print(f'The purchase amount does not seem to impact the rating reviews as the range is quite silm at ${diff_review_usd}.')
```

```python
# label code block section 
print("*"*9)
print("Questions")
print("*"*9)

# output the questions to be answered here
print("Q1) Which season yields the highest amount of purchases overall?\n"
      "Q3) What is the relationship between the customer ages and their frequency of purchases?\n"
      "Q5) Of the most popular sizes purchased overall, which has yields the highest reviews?\n"
      "Q6) How frequently is a promo code used by the customers with the highest frequency of purchases?")
```

```python
print("*"* 12)
print("Q1) Findings")
print("*"* 12)
season_sum = round(shopping_data.groupby('Season')['Purchase Amount (USD)'].sum(),2)

season_max = season_sum.idxmax() # assignment of highest season based on highest USD total
season_min = season_sum.idxmin() # assignment of lowest season based on highest USD total

amount_max = season_sum.max()  # total USD of highest season 
amount_min = season_sum.min()   # total USD of lowest season 
diff_usd = round(amount_max - amount_min, 2)

print(
    f"Q1) The most profitable season is {season_max}, yielding ${amount_max}."
    f"\n This amount is ${diff_usd} more than the lowest purchase amount of ${amount_min} during the lowest yield season of {season_min}.")
```

```python
print("*"*80)
print("Q3) Investigating seasonality of frequency and age with .dropna() and .groupby()") 
print("*"*80)

age_freq = round(shopping_data.dropna(subset=['Frequency of Purchases']).groupby('Frequency of Purchases')['Age'].mean(),1)
age_freq
```

```python
print("*"* 12)
print("Q3) Findings")
print("*"* 12)

age_min = age_freq.min()
age_max = age_freq.max() 
diff_age = round(age_max - age_min, 2) 

# add the commas to the list of column names
freq_cols = ', '.join(age_freq.index.tolist())
print(
    f" Q3) There is not a relationship with frequency of purchases and age across the frequency columns: {freq_cols}."
    f"\n The average age ranges from {age_max} and {age_min}, with a difference of {diff_age} year(s).")
```

```python
print("*"*80)
print("Q5) Investigating seasonality of item size and review rating with .pivot_table()") 
print("*"*80)# "Q5) Of the most popular sizes purchased overall, which  yields the highest reviews?\n"

size_cols = round(shopping_data.dropna(subset=['Review Rating']).pivot_table(values='Review Rating', index='Size', aggfunc= 'mean').sort_values('Review Rating', ascending=False), 2)
size_cols
```

```python
print("*"* 12)
print("Q5) Findings")
print("*"* 12)

size_max = ''.join(size_cols.idxmax().tolist()) #  size with highest average review rating 
size_min = ''.join(size_cols.idxmin().tolist()) #  size with lowest amount of review rating

reviews_max = round(size_cols.max().values[0], 2) # highest average review rating across all sizes
reviews_min = round(size_cols.min().values[0], 2) # lowesr average review rating across all sizes

print(f'The most popular size is {size_max} with an average review rating of {reviews_max}. Oppositely, the least popular size {size_min} has an average review rating of {reviews_min}.')
```
