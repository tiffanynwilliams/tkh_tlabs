## 1. REST in Plain English

REST stands for **Representational State Transfer**. In practice, REST APIs expose resources through URLs and let you interact with them using HTTP methods.

Common HTTP methods:

| Method | Typical Purpose | Example |
|---|---|---|
| `GET` | Read data | Get a user |
| `POST` | Create data | Create a post |
| `PUT` | Replace data | Replace a user profile |
| `PATCH` | Partially update data | Update one field |
| `DELETE` | Delete data | Delete a post |

Common status codes:

| Code | Meaning |
|---|---|
| `200` | OK |
| `201` | Created |
| `204` | No Content |
| `400` | Bad Request |
| `401` | Unauthorized |
| `403` | Forbidden |
| `404` | Not Found |
| `500` | Server Error |

## 6. Error Handling

Real API clients should handle unsuccessful responses. The method `response.raise_for_status()` raises an exception for HTTP error responses.


````python
def get_json(url, params=None):
    response = requests.get(url, params=params, timeout=10)
    response.raise_for_status()
    return response.json()

try:
    data = get_json(f"{BASE_URL}/posts/1")
    pprint(data)
except requests.HTTPError as error:
    print("HTTP error:", error)
except requests.RequestException as error:
    print("Network/request error:", error)
````

## 7. A Basic API Client Class


````python
class APIClient:
    def __init__(self, base_url):
        self.base_url = base_url.rstrip("/")

    def build_url(self, endpoint):
        endpoint = endpoint.lstrip("/")
        return f"{self.base_url}/{endpoint}"

    def get(self, endpoint, params=None):
        url = self.build_url(endpoint)
        response = requests.get(url, params=params, timeout=10)
        response.raise_for_status()
        return response.json()

client = APIClient(BASE_URL)
post = client.get("/posts/1")
pprint(post)
````

## 8. Inheritance and `super()`

Inheritance lets one class reuse and extend another class.

`super()` lets a child class call methods from its parent class.


````python
class AuthenticatedAPIClient(APIClient):
    def __init__(self, base_url, token):
        super().__init__(base_url)
        self.token = token

    def get_headers(self):
        return {
            "Authorization": f"Bearer {self.token}",
            "Accept": "application/json",
        }

    def get(self, endpoint, params=None):
        url = self.build_url(endpoint)
        response = requests.get(
            url,
            params=params,
            headers=self.get_headers(),
            timeout=10,
        )
        response.raise_for_status()
        return response.json()

secure_client = AuthenticatedAPIClient(BASE_URL, token="fake-demo-token")
print(secure_client.build_url("posts/1"))
print(secure_client.get_headers())
````

## 9. Polymorphism

Polymorphism means different classes can share the same interface but implement behavior differently.

Below, each client has a `fetch_resource()` method, but each one fetches a different kind of resource.


````python
class PostsClient(APIClient):
    def fetch_resource(self, resource_id):
        return self.get(f"posts/{resource_id}")

class UsersClient(APIClient):
    def fetch_resource(self, resource_id):
        return self.get(f"users/{resource_id}")

class TodosClient(APIClient):
    def fetch_resource(self, resource_id):
        return self.get(f"todos/{resource_id}")

clients = [
    PostsClient(BASE_URL),
    UsersClient(BASE_URL),
    TodosClient(BASE_URL),
]

for api_client in clients:
    result = api_client.fetch_resource(1)
    print(type(api_client).__name__)
    pprint(result)
    print("-" * 60)
````

## 10. A More Reusable Design

You can create a generic resource client where each subclass only defines the resource name.


````python
class ResourceClient(APIClient):
    resource_name = None

    def __init__(self, base_url):
        super().__init__(base_url)
        if self.resource_name is None:
            raise ValueError("Subclasses must define resource_name")

    def list(self, params=None):
        return self.get(self.resource_name, params=params)

    def retrieve(self, resource_id):
        return self.get(f"{self.resource_name}/{resource_id}")

class CommentsClient(ResourceClient):
    resource_name = "comments"

class AlbumsClient(ResourceClient):
    resource_name = "albums"

comments_client = CommentsClient(BASE_URL)
albums_client = AlbumsClient(BASE_URL)

print("Comment 1")
pprint(comments_client.retrieve(1))

print("
Album 1")
pprint(albums_client.retrieve(1))
````

## 12. Reference Solution for the Challenge

Open this only after attempting the challenge.


````python
class BaseJSONPlaceholderClient:
    def __init__(self, base_url=BASE_URL):
        self.base_url = base_url.rstrip("/")

    def build_url(self, endpoint):
        endpoint = endpoint.lstrip("/")
        return f"{self.base_url}/{endpoint}"

    def request(self, method, endpoint, **kwargs):
        url = self.build_url(endpoint)
        response = requests.request(method, url, timeout=10, **kwargs)
        response.raise_for_status()

        if not response.content:
            return None
        return response.json()

class PostsSDK(BaseJSONPlaceholderClient):
    def __init__(self, base_url=BASE_URL):
        super().__init__(base_url)

    def get_post(self, post_id):
        return self.request("GET", f"posts/{post_id}")

    def list_posts(self, user_id=None):
        params = {}
        if user_id is not None:
            params["userId"] = user_id
        return self.request("GET", "posts", params=params)

    def create_post(self, title, body, user_id):
        payload = {
            "title": title,
            "body": body,
            "userId": user_id,
        }
        return self.request("POST", "posts", json=payload)

class TodosSDK(BaseJSONPlaceholderClient):
    def __init__(self, base_url=BASE_URL):
        super().__init__(base_url)

    def get_todo(self, todo_id):
        return self.request("GET", f"todos/{todo_id}")

    def list_todos(self, user_id=None, completed=None):
        params = {}
        if user_id is not None:
            params["userId"] = user_id
        if completed is not None:
            params["completed"] = str(completed).lower()
        return self.request("GET", "todos", params=params)


def demo_sdk(client):
    if hasattr(client, "list_posts"):
        records = client.list_posts(user_id=1)
        print(f"{type(client).__name__}: {len(records)} posts")
    elif hasattr(client, "list_todos"):
        records = client.list_todos(user_id=1)
        print(f"{type(client).__name__}: {len(records)} todos")
    else:
        raise TypeError("Unsupported client type")

posts_sdk = PostsSDK()
todos_sdk = TodosSDK()

demo_sdk(posts_sdk)
demo_sdk(todos_sdk)

pprint(posts_sdk.create_post("SDK post", "Created from our SDK", 1))
````

## 13. Pagination in REST APIs

Many REST APIs do not return every record in one response. Instead, they return one *page* of results at a time.

Common pagination styles:

1. **Page + limit**: `?page=1&limit=10`
2. **Offset + limit**: `?offset=0&limit=10`
3. **Cursor-based**: `?cursor=abc123&limit=10`
4. **Link header pagination**: response headers include links to the next page

JSONPlaceholder supports a page-like pattern with `_page` and `_limit`.


````python
# Example: get the first page of posts
params = {"_page": 1, "_limit": 5}
response = requests.get(f"{BASE_URL}/posts", params=params, timeout=10)
response.raise_for_status()

page_1_posts = response.json()
print(f"Returned {len(page_1_posts)} posts")
for post in page_1_posts:
    print(post["id"], post["title"])
````

## 14. Looping Through Multiple Pages

A common pattern is to keep requesting pages until the API returns an empty list.

This pattern works for APIs where an empty response means there is no more data.


````python
def fetch_all_posts(limit=20):
    all_posts = []
    page = 1

    while True:
        response = requests.get(
            f"{BASE_URL}/posts",
            params={"_page": page, "_limit": limit},
            timeout=10,
        )
        response.raise_for_status()
        posts = response.json()

        if not posts:
            break

        all_posts.extend(posts)
        print(f"Fetched page {page}: {len(posts)} posts")
        page += 1

    return all_posts

all_posts = fetch_all_posts(limit=25)
print(f"Total posts fetched: {len(all_posts)}")
````

## 15. Pagination with Classes and `super()`

Pagination logic is often reusable. Instead of writing the same loop in every client class, you can put it in a base class.

The subclass can provide resource-specific methods while still reusing the base request and pagination behavior.


````python
class PaginatedResourceClient(ResourceClient):
    def list_page(self, page=1, limit=10, params=None):
        params = params.copy() if params else {}
        params.update({"_page": page, "_limit": limit})
        return self.list(params=params)

    def list_all(self, limit=10, max_pages=10, params=None):
        results = []
        page = 1

        while page <= max_pages:
            items = self.list_page(page=page, limit=limit, params=params)

            if not items:
                break

            results.extend(items)
            page += 1

        return results

class PaginatedPostsClient(PaginatedResourceClient):
    resource_name = "posts"

    def __init__(self, base_url=BASE_URL):
        super().__init__(base_url)

paginated_posts_client = PaginatedPostsClient()
posts = paginated_posts_client.list_all(limit=30, max_pages=5)
print(f"Fetched {len(posts)} posts")
````
